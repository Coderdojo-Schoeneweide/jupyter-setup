# Kluge Züge

Hier zeigen wir dir, wie man den Intelino-Zug mit Python programmieren kann.

Das was du hier siehst ist ein Jupyter Notebook. Du kannst die Code-Zeilen ausführen, indem du sie anklickst und danach Shift+Enter drückst.

```python
import asyncio
from intelino.trainlib_async import TrainScanner
from intelino.trainlib_async.enums import SteeringDecision

print('Jetzt haben wir alle Dinge importiert, die wir für die Züge brauchen! Sehr gut :)')
```

## Zug ID
Als erstes brauchen wir die Zug ID. Diese kannst du auf einem Zettel in der Box deines Zuges finden. Sie sollte ungefähr so aussehen: `XX:XX:XX:XX:XX:XX` (die X sind aber Buchstaben / Zahlen).
Wenn du sie gefunden hast, trage sie in das nächste Kästchen ein:

```python
# WICHTIG: Trage hier die ID von DEINEM Zug ein!
ZUG_ID = "XX:XX:XX:XX:XX:XX"
```

Wenn alles richtig funktioniert hat, dann kannst du den Zug jetzt über die nächste Zeile blinken lassen.

```python
async def main():
    async with TrainScanner(ZUG_ID) as train:
        # wiederhole alles fünf mal
        for i in range(5):
            # mach die Lichter Rot
            await train.set_top_led_color(255, 0, 0)
            
            # warte 0.2 Sekunden
            await asyncio.sleep(0.2)
            
            # mach die Lichter blau
            await train.set_top_led_color(0, 0, 255)
            
            # warte 0.2 Sekunden
            await asyncio.sleep(0.2)

await asyncio.create_task(main())
```

Probier am besten mal aus, wie du die Farben einstellen kannst!

Die Farben werden über die Zahlen in `await train.set_top_led_color(255, 0, 0)` bestimmt.

Unter [https://rgbcolorpicker.com/](https://rgbcolorpicker.com/) kannst due sehen welche Farben wie gemischt werden. Probier mal aus deine eigenen Farbkombinationen zu erschaffen!

## Fahren
Ok, jetzt wollen wir den Zug auch mal fahren lassen!

Dafür kannst du das folgende Kästchen verwenden:

```python
async def main():
    async with TrainScanner(ZUG_ID) as train:
        # losfahren mit 40 Centimetern pro Sekunde
        await train.drive_at_speed(40)

        # eine Sekunde warten
        await asyncio.sleep(1)

        # Stoppe die Fahrt
        await train.stop_driving()

await asyncio.create_task(main())
```

Die wichtigen Stellen sind die folgenden drei Zeilen:

1. `await train.drive_at_speed(40)`:
   Hier sagen wir, dass der Zug losfahren soll. Die `40` sagt wie schnell der Zug fahren soll.
2. `await asyncio.sleep(1)`:
   Hier sagen wir, dass wir eine Sekunde warten wollen. In dieser Zeit fährt der Zug.
3. `await train.stop_driving()`:
   Hier sagen wir, dass der Zug stoppen soll. Er hört jetzt auf zu fahren.


Versuch gerne mal den Code zu erweitern. Wenn du nach `await train.stop_driving()` weitere Zeilen einfügst, werden diese nacheinander ausgeführt. Du kannst auch erstmal probieren den Zug langsamer oder schneller fahren zu lassen. Dafür kannst du die `40` zu einer anderen Zahl ändern.

## Weichen

Wenn du zu einer Weiche kommst, musst du entscheiden, ob du links, geradeaus oder rechts abbiegen möchtest. Dafür benutzen wir `await train.set_next_split_steering_decision(SteeringDecision.RIGHT)`.

1. `await train.set_next_split_steering_decision(SteeringDecision.RIGHT)`: Rechts abbiegen
2. `await train.set_next_split_steering_decision(SteeringDecision.FORWARD)`: Geradeaus fahren
3. `await train.set_next_split_steering_decision(SteeringDecision.LEFT)`: Links abbiegen

Damit legen wir für die nächste Weiche fest, wie wir abbiegen.

```python
async def main():    
    async with TrainScanner(ZUG_ID) as train:
        await train.drive_at_speed(40)

        # wir biegen rechts ab
        await train.set_next_split_steering_decision(SteeringDecision.RIGHT)
        await asyncio.sleep(5)
        await train.stop_driving()

await asyncio.create_task(main())
```

## Wagen abkoppeln

Als letztes zeige ich euch noch, wie man den Wagen abkoppeln kann. Dazu können wir einfach `await train.decouple_wagon()` verwenden.

```python
async def main():    
    async with TrainScanner(ZUG_ID) as train:
        await train.drive_at_speed(40)

        await asyncio.sleep(1)

        # wir biegen rechts ab
        await train.decouple_wagon()
        await asyncio.sleep(1)
        await train.stop_driving()

await asyncio.create_task(main())
```
